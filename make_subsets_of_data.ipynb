```python
import datasets
from datasets import load_dataset
import pandas as pd

```

```python
ds = load_dataset("trl-lib/tldr", split='validation')
```

```python
ds
```

```python
subreddits = []
for i in range(len(ds)):
    subredit_name = ds['prompt'][i]
    start = subredit_name.find('SUBREDDIT: /r') + 14
    end = subredit_name.find('\n', start)
    my_name = subredit_name[start:end]
    subreddits.append(my_name.lower())
    

print(subreddits)
```

```python
topics = []
for my_name in subreddits:
    if my_name not in topics:
        topics.append(my_name)
        
print(topics)
```

```python
len(topics)
```

```python
for i in topics:
    key_word = i
    sub_topics = key_word + "_ds"
    sub_topics = []
    for i in range(len(ds)):
        subredit_name = ds['prompt'][i]
        start = subredit_name.find('SUBREDDIT: /r') + 14
        end = subredit_name.find('\n', start)
        my_name = subredit_name[start:end]
        if(my_name.lower() == key_word):
            sub_topics.append(ds[i])
    print(f"Subset {key_word} has lenght {len(sub_topics)}")
    df_sub_topics = pd.DataFrame(sub_topics)
    df_sub_topics.to_csv('validation_data/%s.csv' % key_word)
```
